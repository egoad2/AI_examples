## What makes a notebook different from a Python script?

A Python script is usually a text file containing code that is executed from top to bottom in one run. A Jupyter notebook combines:

- **Markdown cells** for instructions and formatted explanations;
- **code cells** for Python expressions and statements;
- **outputs** for text, tables, plots, images, or interactive widgets;
- a persistent **kernel state** for variables and imported modules; and
- a visual interface that can rerun individual steps.

Notebooks are especially useful for demonstrations, data exploration, reports, and learning. Scripts remain preferable for production software, automation, repeatable command-line tools, and applications with large test suites.

## 1. How to run this notebook

### The environment

Jupyter has two main parts:

1. **The notebook interface** displays and edits cells.
2. **The kernel** executes Python code and stores variables in memory.

Changing a Markdown or code cell does not change the kernel by itself. The kernel changes only when code is executed.

### Common execution methods

| Action | JupyterLab / VS Code |
|---|---|
| Run the selected cell and move to the next | `Shift+Enter` |
| Run the selected cell and remain in it | `Ctrl+Enter` |
| Insert a cell below | `B` in command mode, or use the toolbar |
| Insert a cell above | `A` in command mode, or use the toolbar |
| Change a cell to code mode | `Y` in command mode |
| Change a cell to Markdown mode | `M` in command mode |
| Open the command palette | `Ctrl+Shift+P` |

> On JupyterLab, press `Esc` to enter command mode; press `Enter` to enter edit mode for the selected cell.

This notebook should be run from top to bottom the first time.

## 2. Code cells

Code cells contain Python. Run the next cell using `Shift+Enter`.

The result is shown directly below the cell. In notebooks, the result of the final expression in a cell is displayed automatically.

````python
# This is a Python comment.
# Comments explain code but are not executed.

2 + 3 * 4
````

The expression above produced the result `14`. Before this demonstration continues, try:

````python
print("This text is printed explicitly.")
print("The print() function is useful for multiple values or progress messages.")

first_name = "Ada"
print("Hello,", first_name)
````

### The difference between printing and final-expression display

Both examples are used in notebooks, but they behave differently:

- `print()` writes text to the output area.
- A final expression is displayed by the notebook interface.
- Intermediate expressions are not displayed unless they are printed or explicitly displayed.

The following cell has two expressions. Only the final expression is shown automatically.

````python
10 + 5
20 + 7
````

Python is executing both expressions, but only the value of the last expression is automatically displayed. To see both values, use `print()` or assign them to variables.

````python
sum_one = 10 + 5
sum_two = 20 + 7

print("sum_one =", sum_one)
print("sum_two =", sum_two)
````

## 3. Variables and notebook state

Variables created in a notebook are stored by the kernel. They persist between code cells and can be reused by later cells.

````python
student_name = "Sam"
course = "Python"
completed_lessons = 3

student_name, course, completed_lessons
````

````python
# Variables from earlier cells are available here.
print(f"{student_name} is learning {course}.")
print("Completed lessons:", completed_lessons)
````

### Increasing a variable

This updates a variable currently stored in the kernel.

````python
completed_lessons = completed_lessons + 1
completed_lessons
````

### A notebook is not always executed in file order

A notebook can display cells top to bottom, but students may execute cells in a different order. Execution numbers show the order in which code was actually run.

For example, if `completed_lessons` is changed, the old value may no longer match the value expected in a previously run cell. This is one of the most common sources of notebook confusion.

## 4. Markdown cells

## 5. Working with lists and dictionaries

Python concepts used elsewhere in a notebook behave exactly as they do in a script. The notebook mainly provides a more flexible way to execute and display them.

````python
temperatures_c = [18, 21, 19, 24, 23]

# Python methods and built-in functions work normally.
average_temperature = sum(temperatures_c) / len(temperatures_c)

print("Temperatures:", temperatures_c)
print("Average:", round(average_temperature, 2), "degrees C")
````

## 6. Functions and reusable analysis

Functions define reusable calculations. Their names remain available in the kernel after they are executed.

````python
def fahrenheit_to_celsius(fahrenheit):
    'Convert a Fahrenheit temperature to Celsius.'
    celsius = (fahrenheit - 32) * 5 / 9
    return celsius


for temperature in (32, 68, 212):
    result = fahrenheit_to_celsius(temperature)
    print(f"{temperature} F = {result:.1f} C")
````

## 8. Notebook line and cell magics

**Magics** are notebook features, not standard Python syntax.

- `%` controls one line.
- `%%` controls an entire cell.

The next examples measure execution time and list variables in the kernel.

````python
import time

%time sum(range(1_000_001))
````

````python
%%time

total = 0
for number in range(1_000_001):
    total += number

print("Total:", total)
````

````python
# %who prints the names of variables currently stored in the kernel.
%who
````

Another useful family of magics can interact with the operating system. Availability depends on the notebook environment:

- `%pwd` shows the notebook server's current working directory.
- `%ls` lists files.
- `%matplotlib inline` asks Jupyter to display Matplotlib plots inside the notebook.
- `!pip install package_name` runs a shell command, but `%pip install package_name` is preferred because it uses the current Jupyter kernel environment.

````python
%pwd
````

## 10. Exceptions are still Python exceptions

A notebook shows a traceback when code fails. Read a traceback from the bottom upward:

1. The final line identifies the exception type and message.
2. Earlier frames show the code path that caused it.
3. The highlighted line identifies where execution stopped.

The next cell intentionally demonstrates an error. Run it, read the traceback, then continue.

````python
names = ["Ada", "Grace", "Linus"]

# The valid indexes are 0, 1, and 2.
names[5]
````

The error is `IndexError: list index out of range`.

A failed cell does not necessarily stop later independent cells. However, a later cell may use variables or functions that the failed cell never created.

````python
# Corrected version
names[2]
````

## 11. Reordering and stale state

The following pattern is a notebook-specific debugging exercise.

````python
# STEP A: Run this cell.
counter = 10
````

````python
# STEP B: Run this cell.
counter = counter + 1
print("Counter:", counter)
````

Try the following:

1. Run **STEP B** several times.
2. Observe that `counter` keeps increasing.
3. Run **STEP A** again to reset it.
4. Run **STEP B** once to verify the result.

This behavior happens because STEP B changes kernel state. It is not caused by a bug in the notebook code.

## 12. Restarting the kernel

Restarting the kernel clears all Python variables and imported modules. It does not delete the notebook text or code cells.

Use **Restart Kernel** when:

- variables have become inconsistent;
- a package or environment was changed;
- the kernel is slow or unresponsive;
- too many experiments have created stale state; or
- the notebook should be demonstrated from a known initial state.

After restarting, the notebook must be run from the beginning. Running only one later cell will not recreate variables, imports, and functions from earlier cells.

## 13. Run-all behavior

**Run All** executes every code cell in notebook order.

Use it to verify that a notebook works from a clean state. Before a Run All demonstration:

1. Save the notebook.
2. Restart the kernel.
3. Clear old outputs if students should start without previous results.
4. Run every code cell from top to bottom.

When debugging, run one cell at a time. When verifying, run all from a clean kernel.

## 14. Saving and sharing

Use the following practices:

- save after meaningful edits;
- use Markdown so that notebook outputs are easy to understand;
- clear outputs containing private or sensitive data;
- include package requirements;
- avoid storing passwords or API keys in cells;
- do not rely on a variable that a notebook runner may not have executed;
- add enough comments and Markdown that the notebook can stand alone.

To share only source code, export the notebook to Python or another supported format. JupyterLab and VS Code usually provide export options in the File menu.
